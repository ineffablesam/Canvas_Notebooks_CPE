```python
# TASK: Run this example, then make the requested change.
# WHY: The output gives you evidence about how the code works.
# CHECK: Compare the result with your prediction before moving on.

def average_one(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

def average_two(numbers):
    total = 0
    for number in numbers:
        total = total + number
    return total / len(numbers)

def average_three(numbers):
    return numbers.sum() / len(numbers)
```
