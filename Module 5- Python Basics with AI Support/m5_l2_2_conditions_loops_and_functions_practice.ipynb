```python
# WORKED EXAMPLE
# Read the word problem first and predict the result before running this cell.
# Every printed value is evidence you can cite in your explanation.

def feedback(score):
    if score >= 80:
        return "ready"
    elif score >= 60:
        return "review"
    return "restart"

for value in [55, 72, 91]:
    print(value, feedback(value))
```
