```python
# WORKED EXAMPLE
# Read the word problem first and predict the result before running this cell.
# Every printed value is evidence you can cite in your explanation.

import pandas as pd

data = pd.DataFrame({
    "team": ["A", "A", "B", "B"],
    "score": [72, 88, 65, 91],
})
filtered = data[data["score"] >= 80]
print(filtered.sort_values("score", ascending=False))
print(data.groupby("team")["score"].mean())
```

```python
# YOUR WORD-PROBLEM INVESTIGATION
# Mission: Add minimum and maximum scores to the grouped report and write one sentence comparing the teams without hiding group size.
# Change one thing at a time, rerun, and keep the output that supports your answer.

summary = (data.groupby("team")
           .agg(mean_score=("score", "mean"), rows=("score", "size"))
           .reset_index())
print(summary)
# Practice: add minimum and maximum scores to the grouped summary.
```
