````python
# TASK: Define the transparent rule-based Event Inspector used before machine learning.
# WHY: This code creates evidence used by this checkpoint and the next project stage.
# CHECK: Trace each condition and verify that the returned score, label, and reasons agree with one another.

def inspect_event(event):
    score = 0
    reasons = []

    if event.get("day_of_week") in ["Friday", "Saturday"]:
        score += 2
        reasons.append("weekend timing")
    if event.get("start_hour") in [17, 18, 19]:
        score += 1
        reasons.append("after-school start")
    if event.get("free_event") is True:
        score += 1
        reasons.append("free admission")
    if event.get("promotion_days", 0) >= 7:
        score += 1
        reasons.append("at least one week of promotion")

    if score >= 4:
        label = "promising"
    elif score >= 2:
        label = "possible"
    else:
        label = "needs review"
    return {"label": label, "score": score, "reasons": reasons}
````

```python
# TASK: Create a separate clean copy and apply the documented normalization and validity rules.
# WHY: This code creates evidence used by this checkpoint and the next project stage.
# CHECK: Never silently overwrite the raw evidence; connect every conversion, replacement, or imputation to a stated rule.

clean = raw.copy()
for column in ["event_type", "organizer_type", "day_of_week", "venue_type",
               "indoor_outdoor", "registration_required", "free_event",
               "food_available", "food_type", "prize_available",
               "email_campaign", "weather_forecast", "actual_weather",
               "campus_activity_level", "high_turnout"]:
    clean[column] = clean[column].astype("string").str.strip().str.lower()

clean["day_of_week"] = clean["day_of_week"].replace({"fri": "friday"})
clean["actual_weather"] = clean["actual_weather"].replace({"rainy": "rain"})
for column in ["registration_required", "free_event", "food_available",
               "prize_available", "email_campaign", "high_turnout"]:
    clean[column] = clean[column].replace({"y": "yes", "n": "no", "true": "yes", "false": "no"})

clean["ticket_price_usd"] = clean["ticket_price_usd"].astype("string").str.replace("$", "", regex=False)
numeric_columns = ["start_hour", "duration_hours", "capacity", "accessibility_score",
                   "ticket_price_usd", "promotion_days", "promotion_channels",
                   "social_posts", "poster_count", "competing_events",
                   "previous_similar_attendance", "actual_attendance", "attendance_rate"]
for column in numeric_columns:
    clean[column] = pd.to_numeric(clean[column], errors="coerce")

clean = clean.drop_duplicates().copy()
clean.loc[~clean["duration_hours"].between(0.5, 12), "duration_hours"] = pd.NA
clean.loc[clean["ticket_price_usd"] < 0, "ticket_price_usd"] = pd.NA
clean.loc[~clean["accessibility_score"].between(1, 5), "accessibility_score"] = pd.NA
clean.loc[~clean["poster_count"].between(0, 200), "poster_count"] = pd.NA

unresolved_outcome = clean["actual_attendance"].isna() | clean["actual_attendance"].lt(0) | clean["actual_attendance"].gt(clean["capacity"])
clean = clean.loc[~unresolved_outcome].copy()
clean["attendance_rate"] = (clean["actual_attendance"] / clean["capacity"]).round(3)
clean["high_turnout"] = clean["attendance_rate"].ge(0.65).map({True: "yes", False: "no"})
clean["target_quality"] = "observed"
print("Clean rows:", len(clean))
```
