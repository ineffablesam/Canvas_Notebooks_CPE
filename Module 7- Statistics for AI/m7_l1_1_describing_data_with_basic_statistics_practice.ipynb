## 1. Challenge briefing

Five basketball players scored similar point totals, but one superstar scored dramatically more. A commentator wants one number describing a typical performance.

### Decision to make

**How do mean, median, range, standard deviation, and IQR tell different parts of the scoring story?**

### Your deliverable

```python
# WORKED EXAMPLE
# Read the word problem first and predict the result before running this cell.
# Every printed value is evidence you can cite in your explanation.

import pandas as pd

values = pd.Series([12, 13, 14, 15, 16, 60])
print("mean:", values.mean())
print("median:", values.median())
print("range:", values.max() - values.min())
print("std:", round(values.std(), 2))
```

## 4. Solve the word problem

Temporarily remove the extreme value, compare every measure, and recommend the most informative center-plus-spread pair.

Keep your changed code and output. Your explanation should answer the decision—not merely say
that the code “worked.”

```python
# YOUR WORD-PROBLEM INVESTIGATION
# Mission: Temporarily remove the extreme value, compare every measure, and recommend the most informative center-plus-spread pair.
# Change one thing at a time, rerun, and keep the output that supports your answer.

q1, q3 = values.quantile([0.25, 0.75])
iqr = q3 - q1
outliers = values[(values < q1 - 1.5*iqr) | (values > q3 + 1.5*iqr)]
print("IQR:", iqr, "outliers:", outliers.tolist())
# Practice: remove 60 temporarily and compare mean, median, and spread.
```
