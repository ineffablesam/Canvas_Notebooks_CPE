```python
# WORKED EXAMPLE
# Read the word problem first and predict the result before running this cell.
# Every printed value is evidence you can cite in your explanation.

from sklearn.metrics import accuracy_score, confusion_matrix

y_true = [0, 0, 0, 0, 1, 1]
baseline = [0, 0, 0, 0, 0, 0]
model_predictions = [0, 0, 0, 1, 1, 0]
print("baseline accuracy:", accuracy_score(y_true, baseline))
print("model accuracy:", accuracy_score(y_true, model_predictions))
print("confusion matrix:\n", confusion_matrix(y_true, model_predictions))
```
