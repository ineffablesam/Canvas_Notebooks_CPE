```python
# WORKED EXAMPLE
# Read the word problem first and predict the result before running this cell.
# Every printed value is evidence you can cite in your explanation.

from sklearn.linear_model import LogisticRegression

X = [[1,0],[2,0],[3,0],[6,1],[7,1],[8,1]]
y = [0,0,0,1,1,1]
model = LogisticRegression().fit(X, y)
examples = [[4,0],[5,1]]
print("labels:", model.predict(examples))
print("probabilities:\n", model.predict_proba(examples))
```
