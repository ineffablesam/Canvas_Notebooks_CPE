```python
# YOUR WORD-PROBLEM INVESTIGATION
# Mission: Print the mistake indices, label each error type, and name one creature condition absent from the training evidence.
# Change one thing at a time, rerun, and keep the output that supports your answer.

for index, (actual, predicted) in enumerate(zip(y_true, y_pred)):
    if actual != predicted:
        kind = "false positive" if predicted == 1 else "false negative"
        print(index, kind)
# Practice: name one additional condition the training data may not cover.
```
